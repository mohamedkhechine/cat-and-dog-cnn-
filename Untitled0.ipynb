```python
#libraries im gonna need
import numpy as np
import tensorflow as  tf
import tensorflow_datasets as tfds
import matplotlib.pyplot as plt
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("bhavikjikadara/dog-and-cat-classification-dataset")

print("Path to dataset files:", path)
```

```python
import os

# list what's in the downloaded path
for item in os.listdir(path):
    print(item)
```

```python
petimages_path = os.path.join(path, 'PetImages')
print(os.listdir(petimages_path))
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    petimages_path, validation_split=0.2, subset="training", seed=123, image_size=(224,224), batch_size=32
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    petimages_path, validation_split=0.2, subset="validation", seed=123, image_size=(224,224), batch_size=32
)
```

```python
class_names = train_ds.class_names
print(class_names)
```

```python
plt.figure(figsize=(10,10))
for images, labels in train_ds.take(1):      # grab one batch
    for i in range(9):                         # first 9 images
        plt.subplot(3, 3, i+1)                 # 3×3 grid
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")
plt.show()
```

```python
print(train_ds.element_spec)
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Input
from tensorflow.keras.models import Model

train_ds = tf.keras.utils.image_dataset_from_directory(
    petimages_path, validation_split=0.2, subset="training", seed=123, image_size=(224,224), batch_size=32)
val_ds = tf.keras.utils.image_dataset_from_directory(
    petimages_path, validation_split=0.2, subset="validation", seed=123, image_size=(224,224), batch_size=32)

def preprocess(image, label):
    return preprocess_input(image), label

train_small = train_ds.take(150).map(preprocess).prefetch(tf.data.AUTOTUNE)
val_small = val_ds.take(30).map(preprocess).prefetch(tf.data.AUTOTUNE)

inputs = Input(shape=(224,224,3))
base = MobileNetV2(include_top=False, weights="imagenet")
base.trainable = False
x = base(inputs, training=False)
x = GlobalAveragePooling2D()(x)
out = Dense(2, activation="softmax")(x)
model = Model(inputs, out)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

history = model.fit(train_small, validation_data=val_small, epochs=4)
```

```python
for images, labels in train_small.take(1):
    print("Pixel min:", float(images.numpy().min()), "max:", float(images.numpy().max()))
    print("Labels:", labels.numpy()[:15])
    break
```

```python
for images, labels in val_small.take(1):
    print("VAL pixel min:", float(images.numpy().min()), "max:", float(images.numpy().max()))
    print("VAL labels:", labels.numpy()[:10])
    break
```

```python
test_loss,test_acc=model.evaluate(val_small)
print("test accuracy",test_acc)
```

```python
for images, labels in val_small.take(1):
    plt.figure(figsize=(10,10))
    for i in range(9):
        plt.subplot(3,3,i+1)
        # un-preprocess for display (reverse the -1..1 back to 0..255)
        img = (images[i].numpy() + 1) * 127.5
        plt.imshow(img.astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")
    plt.show()
    break
```
